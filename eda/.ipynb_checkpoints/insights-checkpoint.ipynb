# Trade Surveillance Intelligence Platform

## Objective

The objective of this analysis is to identify potentially suspicious trading behaviour from a simulated market session.

The workflow follows a surveillance-first approach:

1. Understand market behaviour
2. Analyze trader behaviour
3. Detect suspicious patterns
4. Apply surveillance rules
5. Prioritize investigations
6. Generate actionable findings

The system combines market activity, trader information, stock metadata and external market context to support surveillance investigations.

# Behavioral Pattern Discovery

Before applying surveillance rules, we explore behavioral anomalies present within the market.

The objective is to identify unusual trading characteristics that warrant further investigation.

The following analyses do not conclude misconduct.

Instead, they highlight patterns that surveillance systems typically investigate.

# Pattern 1 - Excessive Cancellation Activity

Frequent order cancellations may indicate attempts to influence market perception without genuine trading intent.

Surveillance teams often use cancellation ratios as an early indicator of potential spoofing or layering behaviour.>  Excessive Cancellation Activity

```python
cancel_pattern = (
    df.assign(
        cancelled=(df["status"]=="CANCEL")
    )
    .groupby("trader_id")
    .agg(
        total_events=("order_id","count"),
        cancelled=("cancelled","sum")
    )
)

cancel_pattern["cancel_ratio"] = (
    cancel_pattern["cancelled"]
    /
    cancel_pattern["total_events"]
)

cancel_pattern = (
    cancel_pattern
    .sort_values(
        "cancel_ratio",
        ascending=False
    )
)

display(cancel_pattern.head(10))
```

### Observation

The cancellation analysis reveals a clear separation between normal traders and a small subset of participants exhibiting elevated cancellation behaviour.

Most traders maintain relatively low cancellation ratios, suggesting genuine trading intent.

A small number of outliers display significantly higher cancellation activity and therefore warrant additional scrutiny.

High cancellation activity alone is not evidence of misconduct, but it is frequently associated with strategies such as spoofing, layering and quote stuffing.

# Pattern 2 - Volume Concentration

High market participation does not necessarily indicate wrongdoing.

However, traders responsible for a disproportionate percentage of total market volume may have greater ability to influence market behaviour.

```python
volume_pattern = (
    df.groupby("trader_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

display(volume_pattern.head(10))
```

### Observation

Trading volume is not evenly distributed across market participants.

A limited number of traders account for a disproportionately large share of overall market activity.

Such concentration does not automatically indicate manipulation; however, traders capable of influencing large portions of market volume represent a higher surveillance priority because their actions can have greater market impact.

These traders become candidates for accumulation, momentum and market influence analysis.

# Pattern 3 - End-of-Day Activity

Many manipulative strategies occur near market close because closing prices affect valuations, benchmarks and reporting obligations.

We investigate participants with unusually large end-of-day activity.

```python
close_window = df[
    df["timestamp"].dt.strftime("%H:%M:%S")
    >= "15:50:00"
]

close_activity = (
    close_window.groupby("trader_id")
    .agg(
        volume=("quantity","sum")
    )
    .sort_values(
        "volume",
        ascending=False
    )
)

display(close_activity.head(10))
```

### Observation

The analysis highlights participants with elevated activity during the final phase of the trading session.

Market-close activity is important because closing prices are commonly used for valuation, benchmarking and reporting purposes.

While elevated participation near market close may be legitimate, unusually large activity during this period is often reviewed for potential close-price influence or benchmark-related manipulation.

# Pattern 4 - Accumulation Behaviour

Aggressive accumulation occurs when participants repeatedly acquire large positions within a short period.

Such activity may be legitimate, but it can also precede momentum-based manipulation strategies.

```python
buy_volume = (
    df[
        (df["status"]=="EXECUTE")
        &
        (df["side"]=="BUY")
    ]
    .groupby("trader_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

display(
    buy_volume.head(10)
)
```

### Observation

Several traders demonstrate sustained buying behaviour throughout the trading session.

Large-scale accumulation may represent legitimate investment activity; however, it may also precede momentum-based strategies where participants attempt to influence market sentiment through aggressive purchasing.

Accumulation patterns are therefore retained for further surveillance review.

# Pattern 5 - Distribution Behaviour

After accumulating positions, manipulative actors may attempt to rapidly exit positions.

Large sell-side activity is therefore monitored separately.

```python
sell_volume = (
    df[
        (df["status"]=="EXECUTE")
        &
        (df["side"]=="SELL")
    ]
    .groupby("trader_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

display(
    sell_volume.head(10)
)
```

### Observation

The sell-side analysis identifies traders responsible for significant distribution activity.

When substantial selling follows prior accumulation, surveillance teams investigate whether positions were established with the intention of benefiting from subsequent market movement.

Distribution patterns become particularly important when combined with accumulation, volume concentration and price movement indicators.

# Pattern 6 - Abnormal Order Book Activity

Market manipulation often begins with unusual order book behavior.

Rather than focusing on executions, surveillance teams examine how traders interact with the order book itself.

Characteristics:

- Frequent order submissions
- Frequent cancellations
- Large order sizes
- Limited executions

Such behavior may indicate attempts to influence perceived market demand or supply.

```python
orderbook_activity = (
    df.groupby("trader_id")
    .agg(
        total_events=("order_id","count"),
        unique_orders=("order_id","nunique"),
        cancellations=("status",
                        lambda x:(x=="CANCEL").sum())
    )
)

orderbook_activity["cancel_ratio"] = (
    orderbook_activity["cancellations"]
    /
    orderbook_activity["total_events"]
)

orderbook_activity = (
    orderbook_activity
    .sort_values(
        "cancel_ratio",
        ascending=False
    )
)

display(orderbook_activity.head(10))
```

### Observation

A small subset of traders generated substantially more order book activity than their peers.

Notably:

- T007
- T018

display elevated cancellation behavior and repeated order submissions.

Such behavior warrants further surveillance analysis.

# Pattern 7 - Buy-Sell Symmetry

Under normal market conditions, buy and sell activity varies significantly across participants.

When a trader repeatedly buys and sells nearly identical quantities of the same security, it may indicate artificial volume generation.

This pattern is commonly investigated as a precursor to wash trading.

```python
symmetry = []

for trader in df["trader_id"].unique():

    trader_exec = df[
        (df["trader_id"] == trader)
        &
        (df["status"] == "EXECUTE")
    ]

    buy_qty = trader_exec[
        trader_exec["side"]=="BUY"
    ]["quantity"].sum()

    sell_qty = trader_exec[
        trader_exec["side"]=="SELL"
    ]["quantity"].sum()

    if buy_qty > 0 and sell_qty > 0:

        similarity = (
            min(buy_qty,sell_qty)
            /
            max(buy_qty,sell_qty)
        )

        symmetry.append({
            "trader_id":trader,
            "buy_volume":buy_qty,
            "sell_volume":sell_qty,
            "similarity":round(similarity,2)
        })

symmetry_df = pd.DataFrame(symmetry)

display(
    symmetry_df.sort_values(
        "similarity",
        ascending=False
    ).head(10)
)
```

### Observation

Most traders exhibit directional behaviour, where buy and sell activity differ according to market conditions.

A small subset of participants demonstrates highly symmetrical buy and sell volumes.

Such behaviour may indicate artificial volume generation rather than genuine exposure management and is therefore reviewed as a potential precursor to wash trading.

# Pattern 8 - Price Impact Correlation

Large trading activity alone is not suspicious.

However, when significant trading activity coincides with observable price movement, surveillance teams investigate whether the trader contributed to the price change.

This analysis evaluates the relationship between participant activity and market movement.

```python
impact = (
    df[
        df["status"]=="EXECUTE"
    ]
    .groupby("trader_id")
    .agg(
        total_volume=("quantity","sum"),
        avg_price=("price","mean")
    )
)

impact = (
    impact.sort_values(
        "total_volume",
        ascending=False
    )
)

display(
    impact.head(10)
)
```

```python
plt.figure(figsize=(10,5))

plt.scatter(
    impact["total_volume"],
    impact["avg_price"]
)

for trader in impact.index:

    plt.annotate(
        trader,
        (
            impact.loc[trader,"total_volume"],
            impact.loc[trader,"avg_price"]
        )
    )

plt.title(
    "Trader Volume vs Average Execution Price"
)

plt.xlabel("Volume")
plt.ylabel("Average Price")

plt.show()
```

### Observation

The analysis identifies participants whose trading activity is associated with significantly higher market participation than their peers.

High participation combined with observable market movement may indicate the ability to influence short-term price dynamics.

These traders become candidates for momentum ignition and pump-and-dump surveillance analysis.

# Rule 1 - Spoofing Detection

Spoofing occurs when large orders are placed and subsequently cancelled without genuine intent to trade.

Objective:

Identify traders attempting to influence market perception through non-bona fide orders.

Indicators:

- Large cancelled volume
- High cancellation count
- Limited execution relative to submitted activity

```python
spoofing = (
    df[df["status"]=="CANCEL"]
    .groupby("trader_id")
    .agg(
        cancelled_orders=("order_id","count"),
        cancelled_volume=("quantity","sum")
    )
    .reset_index()
)

spoofing = spoofing[
    (spoofing["cancelled_orders"] >= 5)
]

spoofing["rule"] = "SPOOFING"

display(spoofing)
```

### Finding

The surveillance engine identified traders exhibiting elevated cancellation activity combined with significant cancelled volume.

Such behaviour may indicate attempts to influence perceived supply and demand without genuine trading intent.

# Rule 2 - Layering Detection

Layering involves placing multiple orders across different levels of the order book and cancelling them before execution.

Objective:

Identify participants attempting to create artificial market depth.

Indicators:

- Multiple cancelled orders
- Large cancelled volume
- Repeated behaviour within the same security

```python
layering_alerts = []

cancel_df = df[df["status"]=="CANCEL"]

for trader in cancel_df["trader_id"].unique():

    trader_data = cancel_df[
        cancel_df["trader_id"] == trader
    ]

    grouped = (
        trader_data
        .groupby(["symbol","side"])
        .agg(
            cancelled_orders=("order_id","count"),
            cancelled_volume=("quantity","sum")
        )
        .reset_index()
    )

    suspicious = grouped[
        (grouped["cancelled_orders"] >= 5)
        &
        (grouped["cancelled_volume"] >= 50000)
    ]

    for _, row in suspicious.iterrows():

        layering_alerts.append({

            "trader_id": trader,
            "symbol": row["symbol"],
            "cancelled_orders": row["cancelled_orders"],
            "cancelled_volume": row["cancelled_volume"]

        })

layering_df = pd.DataFrame(layering_alerts)

display(layering_df)
```

### Finding

Layering activity was observed among a small number of participants displaying repeated cancellation behaviour within the same security.

The pattern suggests potential attempts to influence order book perception rather than execute genuine transactions.

# Rule 3 - Quote Stuffing Detection

Quote stuffing occurs when participants generate excessive order traffic through rapid order submission and cancellation.

Objective:

Identify abnormal order book activity capable of distorting market signals.

Indicators:

- Large number of order events
- Elevated cancellation frequency
- Significant cancellation ratio

```python
quote_stuffing = (
    df.groupby("trader_id")
    .agg(
        total_events=("order_id","count"),
        cancellations=("status",
                        lambda x:(x=="CANCEL").sum())
    )
)

quote_stuffing["cancel_pct"] = (
    quote_stuffing["cancellations"]
    /
    quote_stuffing["total_events"]
)

quote_stuffing = (
    quote_stuffing[
        (quote_stuffing["cancellations"] >= 10)
    ]
    .sort_values(
        "cancel_pct",
        ascending=False
    )
)

display(quote_stuffing)
```

### Finding

A limited subset of traders generated disproportionately high order traffic and cancellation activity.

This behaviour is consistent with quote stuffing indicators and warrants additional investigation.

# Rule 4 - Wash Trading Detection

Wash trading occurs when substantially similar buy and sell activity creates artificial market volume without meaningful change in ownership exposure.

Objective:

Identify potentially artificial trading activity.

Indicators:

- Same trader
- Same security
- Similar buy and sell volume
- Significant total traded volume

Such activity may create a misleading impression of liquidity and market interest.

```python
wash_trade_alerts = []

for trader in df["trader_id"].unique():

    trader_exec = df[
        (df["trader_id"] == trader)
        &
        (df["status"] == "EXECUTE")
    ]

    for symbol in trader_exec["symbol"].unique():

        stock_df = trader_exec[
            trader_exec["symbol"] == symbol
        ]

        buy_qty = stock_df[
            stock_df["side"]=="BUY"
        ]["quantity"].sum()

        sell_qty = stock_df[
            stock_df["side"]=="SELL"
        ]["quantity"].sum()

        total_volume = buy_qty + sell_qty

        if total_volume < 10000:
            continue

        if buy_qty == 0 or sell_qty == 0:
            continue

        similarity = (
            min(buy_qty, sell_qty)
            /
            max(buy_qty, sell_qty)
        )

        if similarity >= 0.90:

            wash_trade_alerts.append({

                "trader_id": trader,
                "symbol": symbol,
                "buy_volume": buy_qty,
                "sell_volume": sell_qty,
                "total_volume": total_volume,
                "similarity": round(similarity,2)

            })

wash_df = pd.DataFrame(
    wash_trade_alerts
)

wash_df
```

### Finding

The surveillance engine identified participants exhibiting highly symmetrical buy and sell activity within the same security.

While legitimate market-making activity may occasionally produce balanced trading behaviour, persistent symmetry combined with meaningful trading volume may indicate artificial volume generation.

These participants are escalated for further surveillance review.

```python
if len(wash_df) > 0:

    plt.figure(figsize=(10,5))

    plt.bar(
        wash_df["trader_id"],
        wash_df["similarity"]
    )

    plt.title("Wash Trading Similarity Score")
    plt.ylabel("Buy/Sell Similarity")
    plt.xlabel("Trader")

    plt.show()

else:
    print("No wash trading candidates detected.")
```

# Rule 5 - Momentum Ignition Detection

Momentum ignition occurs when a participant aggressively accumulates positions, contributes to directional market movement and subsequently exits positions.

Objective:

Identify traders potentially attempting to initiate or amplify short-term momentum.

Indicators:

- Significant accumulation
- Significant distribution
- High market participation

```python
momentum_candidates = []

for trader in df["trader_id"].unique():

    trader_df = df[
        (df["trader_id"] == trader)
        &
        (df["status"] == "EXECUTE")
    ]

    buy_qty = trader_df[
        trader_df["side"]=="BUY"
    ]["quantity"].sum()

    sell_qty = trader_df[
        trader_df["side"]=="SELL"
    ]["quantity"].sum()

    if buy_qty > 50000 and sell_qty > 50000:

        momentum_candidates.append({
            "trader_id": trader,
            "buy_volume": buy_qty,
            "sell_volume": sell_qty
        })

momentum_df = pd.DataFrame(momentum_candidates)

momentum_df
```

### Finding

The surveillance engine identified traders exhibiting both significant accumulation and significant distribution activity.

This behaviour may indicate attempts to generate or capitalize on short-term price momentum and therefore warrants further review.

```python
if len(momentum_df) > 0:

    momentum_df.plot(
        x="trader_id",
        y=["buy_volume","sell_volume"],
        kind="bar",
        figsize=(10,5)
    )

    plt.title("Momentum Ignition Candidates")
    plt.ylabel("Volume")
    plt.show()

else:
    print("No momentum ignition candidates detected.")
```

# Rule 6 - Pump and Dump Detection

Pump and dump schemes involve accumulation of positions followed by substantial distribution after market movement.

Objective:

Identify accumulation-distribution cycles.

Indicators:

- Large buy volume
- Large sell volume
- Balanced entry and exit activity

```python
pump_dump_alerts = []

for trader in df["trader_id"].unique():

    trader_exec = df[
        (df["trader_id"] == trader)
        &
        (df["status"] == "EXECUTE")
    ]

    buy_qty = trader_exec[
        trader_exec["side"]=="BUY"
    ]["quantity"].sum()

    sell_qty = trader_exec[
        trader_exec["side"]=="SELL"
    ]["quantity"].sum()

    if buy_qty > 100000 and sell_qty > 100000:

        ratio = min(
            buy_qty,
            sell_qty
        ) / max(
            buy_qty,
            sell_qty
        )

        if ratio > 0.30:

            pump_dump_alerts.append({

                "trader_id": trader,
                "buy_volume": buy_qty,
                "sell_volume": sell_qty,
                "ratio": round(ratio,2)

            })

pump_dump_df = pd.DataFrame(
    pump_dump_alerts
)

pump_dump_df
```

### Finding

Several participants displayed substantial accumulation followed by meaningful distribution activity.

This behaviour may represent position building followed by position unwinding after favorable market movement.

Such traders are escalated for further surveillance review.

```python
if len(pump_dump_df) > 0:

    pump_dump_df.plot(
        x="trader_id",
        y=["buy_volume","sell_volume"],
        kind="bar",
        figsize=(10,5)
    )

    plt.title("Pump and Dump Candidates")
    plt.ylabel("Volume")
    plt.show()

else:
    print("No pump and dump candidates detected.")
```

# Unified Risk Scoring Engine

Individual surveillance alerts are aggregated into a unified trader risk score.

The objective is not to conclude misconduct but to prioritize investigation efforts.

Each surveillance rule contributes to the trader's overall risk profile.

Higher scores indicate increased surveillance attention.

```python
risk_scores = {}
```

```python
def add_score(trader, score):

    risk_scores[trader] = (
        risk_scores.get(trader, 0)
        + score
    )
```

```python
# Rule Contributions

# Spoofing
for trader in spoofing["trader_id"]:
    add_score(trader, 40)

# Layering
for trader in layering_df["trader_id"].unique():
    add_score(trader, 35)
    
# Quote Stuffing
for trader in quote_stuffing.index:
    add_score(trader, 25)

# Wash Trading
# if len(wash_df) > 0:
#     for trader in wash_df["trader_id"].unique():
#         add_score(trader,35)

# Momentum Ignition
if len(momentum_df) > 0:
    for trader in momentum_df["trader_id"]:
        add_score(trader, 30)

# Pump & Dump
if len(pump_dump_df) > 0:
    for trader in pump_dump_df["trader_id"]:
        add_score(trader, 30)

# volume concatenation
top_volume = (
    df.groupby("trader_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

top3 = top_volume.head(3).index

for trader in top3:
    add_score(trader,20)
    
# close price manipulation
top_close_trader = close_summary.index[0]

add_score(top_close_trader, 15)

```

```python
risk_df = pd.DataFrame(

    list(risk_scores.items()),

    columns=[
        "trader_id",
        "risk_score"
    ]
)

risk_df = risk_df.merge(
    traders,
    on="trader_id",
    how="left"
)

risk_df = risk_df.sort_values(
    "risk_score",
    ascending=False
)

risk_df
```

### Risk Score Interpretation

The risk score represents the combined output of all surveillance rules.

A higher score indicates:

- Multiple rule violations
- Repeated suspicious behaviour
- Greater surveillance priority

Risk scores are used to support triage and investigation prioritization.

```python
plt.figure(figsize=(12,5))

plt.bar(
    risk_df["trader_id"],
    risk_df["risk_score"]
)

plt.title("Trader Risk Ranking")
plt.xlabel("Trader")
plt.ylabel("Risk Score")

plt.show()
```

# Triage Classification

The surveillance engine classifies traders according to investigative priority.

Categories:

- CRITICAL
- HIGH
- MEDIUM
- LOW

This enables investigators to focus resources on the most significant cases first.

```python
def classify(score):

    if score >= 100:
        return "CRITICAL"

    elif score >= 60:
        return "HIGH"

    elif score >= 30:
        return "MEDIUM"

    else:
        return "LOW"
```

```python
risk_df["triage"] = (
    risk_df["risk_score"]
    .apply(classify)
)

risk_df
```

```python
risk_df["triage"].value_counts().plot(
    kind="bar",
    figsize=(8,4)
)

plt.title("Triage Distribution")

plt.show()
```
